# Ejercicio 3

**Teorema $($matriz $P(t))$**:

La matriz de probabilidad de transición $P(t) = [pi,j (t)]$ está dada por
$$
P(t)=\sum_{k=0}^{\infty} e^{-rt}\,\frac{(rt)^k}{k!}\,P^k
$$

$$
M \approx \max\{rt+5\sqrt{rt},\,20\}
$$

```python
import sympy as sp

# Matriz de tasas del ejercicio
matriz_tasas = sp.Matrix([
    [0, 2, 3, 0],
    [4, 0, 2, 0],
    [0, 2, 0, 2],
    [1, 0, 3, 0]
])


# Función para construir la matriz uniformizada
def obtener_matriz_uniformizada(matriz_tasas):

    numero_estados = matriz_tasas.rows

    # Calculamos las tasas de salida de cada estado
    tasas_salida = []
    for i in range(numero_estados):
        tasas_salida.append(sum(matriz_tasas.row(i)))

    # r es la tasa máxima de salida
    tasa_maxima = max(tasas_salida)

    # Construimos la matriz P_hat
    matriz_uniforme = sp.zeros(numero_estados)

    for i in range(numero_estados):
        for j in range(numero_estados):
            if i == j:
                matriz_uniforme[i, j] = 1 - sp.Rational(tasas_salida[i], tasa_maxima)
            else:
                matriz_uniforme[i, j] = sp.Rational(matriz_tasas[i, j], tasa_maxima)

    return matriz_uniforme, tasa_maxima


# Función para aproximar P(t)
def calcular_P_t(tiempo, matriz_uniforme, tasa_maxima):
    M = int(max(tasa_maxima * tiempo + 5 * sp.sqrt(tasa_maxima * tiempo), 20))

    numero_estados = matriz_uniforme.rows
    matriz_transicion = sp.zeros(numero_estados)

    for k in range(M + 1):
        coeficiente_poisson = (
            sp.exp(-tasa_maxima * tiempo)
            * (tasa_maxima * tiempo) ** k
            / sp.factorial(k)
        )

        matriz_transicion += coeficiente_poisson * (matriz_uniforme ** k)

    return sp.N(matriz_transicion, 8)
```


1. Use esta propuesta para calcular $P(0.5), P(1)$ y $P(5)$ para la matriz $R$ del Ejercicio 1.

```python
def inciso_a():
    matriz_uniforme, tasa_maxima = obtener_matriz_uniformizada(matriz_tasas)

    print("Matriz uniformizada P_hat:")
    sp.pprint(matriz_uniforme)

    print("\nTasa máxima r:")
    print(tasa_maxima)

    P_05 = calcular_P_t(sp.Rational(1, 2), matriz_uniforme, tasa_maxima)
    P_1 = calcular_P_t(1, matriz_uniforme, tasa_maxima)
    P_5 = calcular_P_t(5, matriz_uniforme, tasa_maxima)

    print("\nP(0.5) =")
    sp.pprint(P_05)

    print("\nP(1) =")
    sp.pprint(P_1)

    print("\nP(5) =")
    sp.pprint(P_5)

    return P_05, P_1, P_5, matriz_uniforme, tasa_maxima

P_05, P_1, P_5, matriz_uniforme, tasa_maxima = inciso_a()
```

2. ¿Se verifica la ecuación de Chapman-Kolmogorov $P(1) = P(0.5)P(0.5)$

```python
def inciso_b(P_05, P_1):

    producto_CK = P_05 * P_05

    print("\nP(0.5)P(0.5) =")
    sp.pprint(sp.N(producto_CK, 8))

    diferencia = P_1 - producto_CK

    error_maximo = max(
        abs(diferencia[i, j])
        for i in range(diferencia.rows)
        for j in range(diferencia.cols)
    )

    print("\nError máximo:")
    print(sp.N(error_maximo, 8))

inciso_b(P_05, P_1)
```

<p align="justify">Como el error máximo obtenido fue del orden de $10^{-7}$, se concluye que $P(1)\approx P(0.5)P(0.5)$, por lo que la ecuación de $\textbf{Chapman-Kolmogorov}$ se cumple satisfactoriamente y la aproximación numérica utilizada resulta adecuada.</p>

# Ejercicio 4

**Teorema (Cotas de error)** para $P(t)$: Para un $t ≥ 0$ fijo, sea:

$$
P^M(t)=\left[p_{ij}^{M}(t)\right]
=\sum_{k=0}^{M} e^{-rt}\,\frac{(rt)^k}{k!}\,P^k
$$

entonces
$$
\left|p_{ij}(t)-p_{ij}^{M}(t)\right|
\leq
\sum_{k=M+1}^{\infty}
e^{-rt}\,\frac{(rt)^k}{k!}
$$

Este teorema se puede usar así:

Suponga que se desea calcular
$P(t)$
con una tolerancia ε. Elija $M$ tal que
$$\sum_{k=M+1}^{\infty}
e^{-rt}\,\frac{(rt)^k}{k!} \leq ϵ$$

Y se puede implementar de acuerdo al siguiente algoritmo de uniformización para $P(t):$

1. Dados $R, t, 0 < ε < 1$
2. Calcular $r$ usando la igualdad en la definición.
3. Calcular $\hat{P}$.
4. $A = \hat{P}; B = e^{−rt}I; c = e^{−rt}; sum = c; k = 1$
5. Mientras $sum < 1 − ε$ hacer:
    *   $A = A\hat{P}$
    *   $c = c ∗ (rt)/k$
    *   $B = B + cA$
    *   $sum = sum + c$
    *   $k = k + 1$
6. $B$ está a ε de $P(t).$

```python
import sympy as sp

def uniformizacion(R, t, eps=1e-5):
    n = R.rows  # número de estados

    # ri contiene las tasas de salida de cada estado
    ri = [sum(R.row(i)) for i in range(n)]

    # r es la mayor tasa de salida
    r = max(ri)

    P_hat = sp.zeros(n)

    for i in range(n):
        for j in range(n):

            # Elementos diagonales
            if i == j:
                P_hat[i, j] = 1 - sp.Rational(ri[i], r)

            # Elementos fuera de la diagonal
            else:
                P_hat[i, j] = sp.Rational(R[i, j], r)

    # Algoritmo de uniformización
    A = sp.eye(n)   # A almacenará las potencias sucesivas de P_hat

    # Primer coeficiente de Poisson: e^{-rt}
    c = float(sp.exp(-r * t))

    # Inicializamos la suma:
    B = c * A   # B = e^{-rt} I

    # Probabilidad acumulada de Poisson
    suma = c

    # Contador de términos
    k = 1

    # Se agregan términos hasta que la cola de la serie sea menor que la tolerancia eps
    while (1 - suma) > eps:

        # Calcula P_hat^k de forma recursiva
        A = A * P_hat

        # Actualiza el coeficiente de Poisson
        c = c * (r * t) / k

        # Agrega el nuevo término a la serie
        B = B + c * A

        # Actualiza la probabilidad acumulada
        suma += c

        # Incrementa el contador
        k += 1

    # Devuelve la aproximación de P(t) y el número de términos utilizados
    return sp.N(B, 10), k - 1
```

Repita el ejercicio 3 aplicando este algoritmo con una tolerancia $ε = 0.00001 $ (indique el valor correspondiente de M en cada caso). Compare los resultados.

```python
R = sp.Matrix([
    [0,2,3,0],
    [4,0,2,0],
    [0,2,0,2],
    [1,0,3,0]
])

P05, M05 = uniformizacion(R, 0.5)
P1, M1 = uniformizacion(R, 1)
P5, M5 = uniformizacion(R, 5)

print("M para t=0.5 =", M05)
print("M para t=1 =", M1)
print("M para t=5 =", M5)

print("\nP(0.5)")
sp.pprint(P05)

print("\nP(1)")
sp.pprint(P1)

print("\nP(5)")
sp.pprint(P5)
```

<p align="justify">Por lo tanto, el algoritmo implementado resulta adecuado para aproximar la matriz de transición $P(t)$. Además, para $t=5$ se observa que la cadena se encuentra muy cercana a su distribución estacionaria.</p>
